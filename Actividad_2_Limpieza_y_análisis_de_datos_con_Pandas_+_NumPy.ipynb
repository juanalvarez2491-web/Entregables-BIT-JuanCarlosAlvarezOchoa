** Identificar las columnas con valores faltantes y aplica una estrategia de tratamiento adecuada.**

```python
df["Animal type"] = (
    df["Animal type"]
    .str.strip()
    .str.lower()
    .str.title()
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("™", "", regex=False)
    .str.replace("?", "", regex=False)
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("European Bisson", "European Bison", regex=False)
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("Red Squirel", "Red Squirrell", regex=False)
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("Red Squierrel", "Red Squirrel", regex=False)
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("Red Squirrell", "Red Squirrel", regex=False)
)
```
