** Identificar las columnas con valores faltantes y aplica una estrategia de tratamiento adecuada.**

```python
df["Animal type"] = (
    df["Animal type"]
    .str.strip()
    .str.lower()
    .str.title()
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("™", "", regex=False)
    .str.replace("?", "", regex=False)
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("European Bisson", "European Bison", regex=False)
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("Red Squirel", "Red Squirrell", regex=False)
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("Red Squierrel", "Red Squirrel", regex=False)
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("Red Squirrell", "Red Squirrel", regex=False)
)
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("European Buster", "European Bison", regex=False)
)
```

```python
df["Country"] = (
    df["Country"]
    .str.strip()
    .str.lower()
    .str.title()
)
```

```python
df["Country"] = (
    df["Country"]
    .str.replace("Pl", "Poland", regex=False)
)
```

```python
df["Country"] = (
    df["Country"]
    .str.replace("De", "Germany", regex=False)
)
```

```python
df["Country"] = (
    df["Country"]
    .str.replace("Cc", "Cz", regex=False)
)
```

```python
df["Country"] = df["Country"].replace({
    "Cz": "Czech",
    "Czech Republic": "Czech"
})
```

**4. Análisis con groupby()**

Una vez que hayas terminado la limpieza, utiliza groupby() para responder preguntas sobre los datos.

Debes realizar al menos 4 análisis diferentes.


 **• Cuál es el número de observaciones por tipo de animal?**

```python
df.groupby("Animal type").size()
```


**• Cuál es el promedio de una variable numérica según el tipo de animal?**

```python
promedio_tipo_animal = df.groupby("Animal type")["Body Length cm"].mean()
promedio_tipo_animal
```

**• Cuál es el promedio de una variable numérica según el país?**

```python
promedio_Peso_Por_Pais = df.groupby("Country")["Weight kg"].mean()
promedio_Peso_Por_Pais
```

• Cuál es el promedio de una variable numérica según el país?

```python
promedio_Peso_Por_Sexo = df.groupby("Gender")["Weight kg"].mean()
promedio_Peso_Por_Sexo
```
